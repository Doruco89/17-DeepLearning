```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

DATA = Path("data/garbage_classification")
OUT = Path("outputs/presentation")

plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

PAPER = "#F7F8F5"
TITLE = "#1E3A28"
CLAY = "#B3562F"
FOREST = "#3D7A45"
GREY = "#5B6660"

OUT.mkdir(parents=True, exist_ok=True)
print("데이터 경로 확인:", DATA.exists())


def grid(items, title, subtitle, filename, ncol=3, cap_color=CLAY):
    missing = [p for p, _ in items if not (DATA / p).exists()]
    if missing:
        print("[건너뜀]", filename, "— 없는 파일:", missing)
        return

    nrow = -(-len(items) // ncol)
    cell_h = 3.4
    header = 1.15
    fig_h = cell_h * nrow + header
    fig_w = 3.0 * ncol

    fig, axes = plt.subplots(nrow, ncol, figsize=(fig_w, fig_h))
    fig.patch.set_facecolor(PAPER)
    axes = axes.ravel() if len(items) > 1 else [axes]

    for ax, (p, cap) in zip(axes, items):
        ax.imshow(Image.open(DATA / p).convert("RGB"))
        ax.set_title(cap, fontsize=11, pad=8, color=cap_color, fontweight="bold")
        ax.set_xticks([])
        ax.set_yticks([])
        for sp in ax.spines.values():
            sp.set_color("#D8E0D4")
            sp.set_linewidth(1.2)

    for ax in axes[len(items):]:
        ax.axis("off")

    fig.subplots_adjust(top=1 - header / fig_h, left=0.03, right=0.97,
                        bottom=0.03, wspace=0.12, hspace=0.28)
    fig.text(0.5, 1 - 0.34 / fig_h, title, ha="center", va="center",
             fontsize=17, fontweight="bold", color=TITLE)
    fig.text(0.5, 1 - 0.72 / fig_h, subtitle, ha="center", va="center",
             fontsize=11.5, color=GREY)

    fig.savefig(OUT / filename, dpi=180, facecolor=PAPER)
    plt.close(fig)
    print("[저장]", filename, f"({len(items)}장)")


label_items = [
    ("clothes/clothes3062.jpg", "clothes3062  →  shoes"),
    ("clothes/clothes3330.jpg", "clothes3330  →  shoes"),
    ("clothes/clothes5002.jpg", "clothes5002  →  shoes"),
    ("clothes/clothes3545.jpg", "clothes3545  →  shoes"),
    ("clothes/clothes2197.jpg", "clothes2197  →  shoes"),
    ("shoes/shoes1826.jpg", "shoes1826  →  clothes"),
]
grid(label_items,
     "폴더가 알려주지 않는 라벨 오류",
     "clothes 폴더에 들어 있던 신발 사진들 · 마지막 한 장은 반대 방향(반바지)",
     "img_label_errors.png")


dup_items = [
    ("green-glass/green-glass2.jpg", "green-glass2"),
    ("green-glass/green-glass292.jpg", "green-glass292"),
    ("green-glass/green-glass443.jpg", "green-glass443"),
]
grid(dup_items,
     "MD5로는 잡히지 않는 근접 중복",
     "같은 그룹으로 묶인 세 장 · 파일 바이트는 서로 다르다",
     "img_near_dup.png", cap_color=FOREST)


fp_items = [
    ("trash/trash357.jpg", "trash357"),
    ("trash/trash384.jpg", "trash384"),
    ("trash/trash413.jpg", "trash413"),
]
grid(fp_items,
     "알고리즘이 같다고 판정한 오탐",
     "흰 배경 + 세로 물체 구도가 같을 뿐, 서로 다른 칫솔 제품이다",
     "img_false_positive.png", cap_color=FOREST)


print("\n완료 —", OUT.resolve())
```
